```python
#Cleaning years - dropping missing values from year, power, fuel_type, transmission, make columns. 
auto24_clean = auto24.dropna(subset=["year", "power_ps", "fuel_type", "transmission", "make"])

#Converting year float to int and keeping only years between 2005-2027
auto24_clean["year"] = auto24_clean["year"].astype(int)
auto24_clean = auto24_clean.loc[(auto24_clean["year"] < 2027) & (auto24_clean["year"] > 2005)]

#Cleaning body types - most of the ads about non-passenger cars are classified as other (trucks, bulldozers, motorcycles, boats etc.)
auto24_clean = auto24_clean.loc[(auto24_clean["body_type"] != "other")]

#cleaning up prices - everything between 500 - 100 000 EUR is kept.
auto24_clean = auto24_clean.loc[(auto24_clean["price_eur"] > 500) & (auto24_clean["price_eur"] < 100000)]


```
